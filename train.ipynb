```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import accuracy_score, log_loss
from xgboost import XGBClassifier

# ===============================
# 1. LOAD DATASET
# ===============================
df = pd.read_csv("tomato_sensor_dataset_500.csv")

X = df.drop("Label", axis=1)
y = df["Label"]

# ===============================
# 2. ENCODE LABELS
# ===============================
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# ===============================
# 3. SCALE FEATURES
# ===============================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# ===============================
# 4. TRAIN TEST SPLIT
# ===============================
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_encoded, test_size=0.2, random_state=42
)

# ===============================
# 5. DEFINE MODEL
# ===============================
model = XGBClassifier(
    n_estimators=1,
    learning_rate=0.1,
    max_depth=4,
    objective="multi:softprob",
    num_class=3,
    use_label_encoder=False,
    eval_metric="mlogloss"
)

# ===============================
# 6. TRAIN FOR 20 EPOCHS
# ===============================
train_accuracy = []
test_accuracy = []
train_loss = []
test_loss = []

for epoch in range(20):
    
    model.n_estimators += 1
    model.fit(X_train, y_train, verbose=False)
    
    # Predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    y_train_prob = model.predict_proba(X_train)
    y_test_prob = model.predict_proba(X_test)
    
    # Accuracy
    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc = accuracy_score(y_test, y_test_pred)
    
    train_accuracy.append(train_acc)
    test_accuracy.append(test_acc)
    
    # Loss
    train_l = log_loss(y_train, y_train_prob)
    test_l = log_loss(y_test, y_test_prob)
    
    train_loss.append(train_l)
    test_loss.append(test_l)
    
    print(f"Epoch {epoch+1}/20 - "
          f"Train Acc: {train_acc:.4f}, "
          f"Test Acc: {test_acc:.4f}, "
          f"Train Loss: {train_l:.4f}, "
          f"Test Loss: {test_l:.4f}")

# ===============================
# 7. SAVE MODEL
# ===============================
joblib.dump(model, "quantum_xgboost_model.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(le, "label_encoder.pkl")

print("Model Saved Successfully!")

# ===============================
# 8. PLOT ACCURACY GRAPH
# ===============================
plt.figure()
plt.plot(range(1, 21), train_accuracy)
plt.plot(range(1, 21), test_accuracy)
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Epoch")
plt.show()

# ===============================
# 9. PLOT LOSS GRAPH
# ===============================
plt.figure()
plt.plot(range(1, 21), train_loss)
plt.plot(range(1, 21), test_loss)
plt.xlabel("Epoch")
plt.ylabel("Log Loss")
plt.title("Loss vs Epoch")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from xgboost import XGBClassifier


# ============================================================
# 1. LOAD YOUR DATASET
# ============================================================

DATASET = "tomato_sensor_dataset_500.csv"

df = pd.read_csv(DATASET)

print("\n========================================")
print("DATASET INFORMATION")
print("========================================")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
print(df.head())

print("\nLabel Distribution:")
print(df["Label"].value_counts())


# ============================================================
# 2. REMOVE EMPTY VALUES
# ============================================================

df = df.dropna()

print("\nDataset after removing missing values:", df.shape)


# ============================================================
# 3. INPUT FEATURES AND TARGET
# ============================================================

features = [
    "Temperature_C",
    "Humidity_percent",
    "Rain",
    "LDR_percent",
    "Moisture_percent"
]

X = df[features]

y = df["Label"]


# ============================================================
# 4. ENCODE LABELS
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\n========================================")
print("LABEL ENCODING")
print("========================================")

for class_name, class_number in zip(
    label_encoder.classes_,
    range(len(label_encoder.classes_))
):
    print(class_name, "=", class_number)


# ============================================================
# 5. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("\n========================================")
print("DATA SPLIT")
print("========================================")

print("Training samples :", len(X_train))
print("Testing samples  :", len(X_test))


# ============================================================
# 6. FEATURE SCALING
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)


# ============================================================
# 7. XGBOOST MODEL
# ============================================================

model = XGBClassifier(
    n_estimators=20,
    learning_rate=0.1,
    max_depth=4,
    min_child_weight=1,
    subsample=0.9,
    colsample_bytree=0.9,

    objective="multi:softprob",
    num_class=3,

    eval_metric="mlogloss",

    random_state=42,
    n_jobs=-1
)


# ============================================================
# 8. TRAIN MODEL
# ============================================================

print("\n========================================")
print("TRAINING XGBOOST MODEL")
print("========================================")

model.fit(
    X_train_scaled,
    y_train,

    eval_set=[
        (X_train_scaled, y_train),
        (X_test_scaled, y_test)
    ],

    verbose=False
)

print("Training completed successfully!")


# ============================================================
# 9. GET TRAINING HISTORY
# ============================================================

results = model.evals_result()

train_loss = results["validation_0"]["mlogloss"]

test_loss = results["validation_1"]["mlogloss"]


# ============================================================
# 10. CALCULATE ACCURACY FOR EACH BOOSTING ROUND
# ============================================================

train_accuracy = []

test_accuracy = []

epochs = range(1, 21)


for epoch in epochs:

    train_probability = model.predict_proba(
        X_train_scaled,
        iteration_range=(0, epoch)
    )

    test_probability = model.predict_proba(
        X_test_scaled,
        iteration_range=(0, epoch)
    )


    train_prediction = np.argmax(
        train_probability,
        axis=1
    )

    test_prediction = np.argmax(
        test_probability,
        axis=1
    )


    train_acc = accuracy_score(
        y_train,
        train_prediction
    )

    test_acc = accuracy_score(
        y_test,
        test_prediction
    )


    train_accuracy.append(train_acc)

    test_accuracy.append(test_acc)


    print(
        f"Epoch {epoch:02d}/20 | "
        f"Train Accuracy: {train_acc*100:.2f}% | "
        f"Test Accuracy: {test_acc*100:.2f}% | "
        f"Train Loss: {train_loss[epoch-1]:.4f} | "
        f"Test Loss: {test_loss[epoch-1]:.4f}"
    )


# ============================================================
# 11. FINAL PREDICTION
# ============================================================

y_pred = model.predict(X_test_scaled)


# ============================================================
# 12. FINAL ACCURACY
# ============================================================

final_accuracy = accuracy_score(
    y_test,
    y_pred
)

print("\n========================================")
print("FINAL MODEL RESULT")
print("========================================")

print(
    f"Test Accuracy: {final_accuracy * 100:.2f}%"
)


# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

print("\n========================================")
print("CLASSIFICATION REPORT")
print("========================================")

print(
    classification_report(
        y_test,
        y_pred,
        target_names=label_encoder.classes_
    )
)


# ============================================================
# 14. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred
)

print("\n========================================")
print("CONFUSION MATRIX")
print("========================================")

print(cm)


plt.figure(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

disp.plot()

plt.title("XGBoost Confusion Matrix")

plt.tight_layout()

plt.savefig(
    "confusion_matrix.png",
    dpi=300
)

plt.show()


# ============================================================
# 15. ACCURACY VS EPOCH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    epochs,
    train_accuracy,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    epochs,
    test_accuracy,
    marker="o",
    label="Testing Accuracy"
)

plt.xlabel("Boosting Round")

plt.ylabel("Accuracy")

plt.title(
    "XGBoost Training and Testing Accuracy"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "accuracy_vs_epoch.png",
    dpi=300
)

plt.show()


# ============================================================
# 16. LOSS VS EPOCH
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    epochs,
    train_loss,
    marker="o",
    label="Training Loss"
)

plt.plot(
    epochs,
    test_loss,
    marker="o",
    label="Testing Loss"
)

plt.xlabel("Boosting Round")

plt.ylabel("Log Loss")

plt.title(
    "XGBoost Training and Testing Loss"
)

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.savefig(
    "loss_vs_epoch.png",
    dpi=300
)

plt.show()


# ============================================================
# 17. SAVE TRAINED MODEL
# ============================================================

joblib.dump(
    model,
    "quantum_xgboost_model.pkl"
)

joblib.dump(
    scaler,
    "scaler.pkl"
)

joblib.dump(
    label_encoder,
    "label_encoder.pkl"
)


print("\n========================================")
print("MODEL FILES SAVED")
print("========================================")

print("quantum_xgboost_model.pkl")
print("scaler.pkl")
print("label_encoder.pkl")


# ============================================================
# 18. TEST YOUR SENSOR VALUE
# ============================================================

sensor_data = pd.DataFrame({

    "Temperature_C": [29.4],

    "Humidity_percent": [68.0],

    "Rain": [24],

    "LDR_percent": [73],

    "Moisture_percent": [58]

})


sensor_scaled = scaler.transform(
    sensor_data[features]
)


prediction = model.predict(
    sensor_scaled
)


probability = model.predict_proba(
    sensor_scaled
)


predicted_label = label_encoder.inverse_transform(
    prediction
)


print("\n========================================")
print("LIVE SENSOR PREDICTION")
print("========================================")

print("Temperature :", 29.4)
print("Humidity    :", 68.0)
print("Rain        :", 24)
print("LDR         :", 73)
print("Moisture    :", 58)

print(
    "\nPredicted Label:",
    predicted_label[0]
)


print("\nPrediction Probability:")

for label, prob in zip(
    label_encoder.classes_,
    probability[0]
):

    print(
        f"{label}: {prob*100:.2f}%"
    )


print("\n========================================")
print("PROGRAM COMPLETED")
print("========================================")
```
